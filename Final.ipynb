## Exploratory Data Analysis

Exploratory Data Analysis (EDA) is performed to understand the structure, distribution, and patterns present in the dataset.

The analysis focuses on customer age, gender, annual income, spending score, and the relationship between annual income and spending score

```python
plt.figure(figsize=(8,5))
plt.hist(df["Age"])
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Annual Income (k$)"])
plt.xlabel("Annual Income (k$)")
plt.ylabel("Number of Customers")
plt.title("Annual Income Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Annual Income (k$)"])
plt.xlabel("Annual Income (k$)")
plt.ylabel("Number of Customers")
plt.title("Annual Income Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Spending Score (1-100)"])
plt.xlabel("Spending Score")
plt.ylabel("Number of Customers")
plt.title("Spending Score Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Annual Income (k$)"],
    df["Spending Score (1-100)"]
)

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Annual Income vs Spending Score")

plt.show()
```

The scatter plot shows the relationship between annual income and spending score.

The visualization indicates that customers have different combinations of income and spending behavior. These differences provide a basis for applying clustering techniques to identify customer segments

```python
X = df[["Annual Income (k$)","Spending Score (1-100)"]]
```

```python
inertia = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

print(inertia)
```

```python
plt.plot(range(2,11),inertia,marker="o")
plt.xlabel("Number of clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```
